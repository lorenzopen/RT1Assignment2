```python
import rospy
import actionlib
import ipywidgets as widgets
from IPython.display import display
from geometry_msgs.msg import Pose, Twist
from nav_msgs.msg import Odometry
from assignment_2_2024.msg import PlanningAction, PlanningGoal
from assignment2.srv import GetLastGoal

import math

class JupyterActionClient:
    def __init__(self):
        
        rospy.init_node('jupyter_action_client', anonymous=True)

        self.x_input = widgets.FloatText(description="X:")
        self.y_input = widgets.FloatText(description="Y:")
        self.send_goal_button = widgets.Button(description="Send Goal", button_style='success')
        self.cancel_goal_button = widgets.Button(description="Cancel Goal", button_style='warning')
        self.prev_goal_button = widgets.Button(description="Get Previous Goal", button_style='info')

        self.position_label = widgets.Label(value="Current Position: x=0.00, y=0.00")
        self.distance_label = widgets.Label(value="Distance from Origin: 0.00 m")

        self.current_pose = Pose()
        self.current_velocity = Twist()

        self.send_goal_button.on_click(self.send_goal)
        self.cancel_goal_button.on_click(self.cancel_goal)
        self.prev_goal_button.on_click(self.get_previous_goal)

        self.ui = widgets.VBox([
            widgets.HBox([self.x_input, self.y_input]),
            self.send_goal_button,
            self.cancel_goal_button,
            self.prev_goal_button,
            self.position_label,
            self.distance_label
        ])

        self.client = actionlib.SimpleActionClient('/reaching_goal', PlanningAction)
        self.client.wait_for_server()

        self.sub_odom = rospy.Subscriber('/odom', Odometry, self.odom_callback)

    def odom_callback(self, msg):
        self.current_pose = msg.pose.pose
        self.current_velocity = msg.twist.twist

        self.update_position()
        self.update_distance()

    def update_position(self):
        x = self.current_pose.position.x
        y = self.current_pose.position.y
        self.position_label.value = f"Current Position: x = {x:.2f}, y = {y:.2f}"

    def update_distance(self):
        x = self.current_pose.position.x
        y = self.current_pose.position.y
        distance = math.sqrt(x**2 + y**2)
        self.distance_label.value = f"Distance from Origin: {distance:.2f} m"

    def send_goal(self, b):
        x = self.x_input.value
        y = self.y_input.value
        goal = PlanningGoal()
        goal.target_pose.pose.position.x = x
        goal.target_pose.pose.position.y = y
        self.client.send_goal(goal)
        rospy.loginfo(f"Sent goal: x = {x}, y = {y}")

    def cancel_goal(self, b):
        self.client.cancel_goal()
        rospy.loginfo("Goal Cancelled")

    def get_previous_goal(self, b):
        rospy.wait_for_service('get_last_goal')
        try:
            get_last_goal = rospy.ServiceProxy('get_last_goal', GetLastGoal)
            response = get_last_goal()
            if response.last_goal:
                rospy.loginfo(f"Previous goal: x={response.last_goal[0]}, y={response.last_goal[1]}")
                self.x_input.value = response.last_goal[0]
                self.y_input.value = response.last_goal[1]
            else:
                rospy.loginfo("No previous goal available.")
        except rospy.ServiceException as e:
            rospy.logerr(f"Service call failed: {e}")

    def display(self):
        display(self.ui)

client = JupyterActionClient()
client.display()
```

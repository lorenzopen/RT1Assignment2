```python
import rospy
import ipywidgets as widgets
from IPython.display import display
from assignment2.srv import GetLastGoal

class PreviousGoalRetriever:
    def __init__(self):
        rospy.init_node('previous_goal_retriever', anonymous=True)

        self.retrieve_button = widgets.Button(description="Retrieve Previous Goal", button_style='info')
        self.result_label = widgets.Label(value="Previous goal will appear here.")

        self.retrieve_button.on_click(self.retrieve_previous_goal)

        self.ui = widgets.VBox([
            self.retrieve_button,
            self.result_label
        ])

    def retrieve_previous_goal(self, b):

        rospy.wait_for_service('get_last_goal')
        try:
            get_last_goal = rospy.ServiceProxy('get_last_goal', GetLastGoal)
            response = get_last_goal()
            if response.last_goal:
                x, y = response.last_goal
                self.result_label.value = f"Previous Goal: x = {x:.2f}, y = {y:.2f}"
                rospy.loginfo(f"Previous goal: x={x}, y={y}")
            else:
                self.result_label.value = "No previous goal available."
                rospy.loginfo("No previous goal available.")
        except rospy.ServiceException as e:
            self.result_label.value = f"Service call failed: {e}"
            rospy.logerr(f"Service call failed: {e}")

    def display(self):
        display(self.ui)

prev_goal_ret = PreviousGoalRetriever()
prev_goal_ret.display()
```
